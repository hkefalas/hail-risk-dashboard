```python
# --- generate_radar.py ---
import pandas as pd
from datetime import datetime
from datetime import timedelta
import requests
import os
from pathlib import Path

def parse_event_times(df, filepath):
    # Extract date from filename
    event_date_str = Path(filepath).stem  # '2025-07-12'
    print(event_date_str)
    event_date = datetime.strptime(event_date_str, "%Y-%m-%d")
    print(event_date)
    # Ensure 'Time' column is string with 4 digits
    df["Time"] = df["Time"].astype(str).str.zfill(4)

    event_times = []
    for time_str in df["Time"]:
        try:
            hour = int(time_str[:2])
            minute = int(time_str[2:])
            dt = event_date.replace(hour=hour, minute=minute)

            # If time is before noon, assume it's UTC and may roll into next calendar day
            if hour < 12:
                dt += timedelta(days=1)

            event_times.append(dt)
        except Exception:
            event_times.append(pd.NaT)

    df["Event Time"] = event_times
    df = df.dropna(subset=["Event Time"])
    return df
        

def get_hail_events_from_csv(csv_path: str, station="DMX") -> pd.DataFrame:
    df = pd.read_csv(csv_path)
    records = []
    for _, row in df.iterrows():
        time_str = str(row.get("Time (UTC)", "")).zfill(4)
        time_obj = parse_utc_time(time_str)
        radar_url = generate_radar_url(time_obj, station=station)

        records.append({
            "Time (UTC)": time_str,
            "Location": row.get("Location", ""),
            "Size (in)": row.get("Size", ""),
            "State": row.get("State", ""),
            "RadarURL": radar_url
        })
    return pd.DataFrame(records)
```

```python
from geopy.distance import geodesic

# Find closest station to a lat/lon
def find_closest_station(lat, lon, stations_df):
    point = (lat, lon)
    stations_df["distance_km"] = stations_df.apply(
        lambda row: geodesic(point, (row["lat"], row["lon"])).km, axis=1
    )
    return stations_df.sort_values("distance_km").iloc[0]

# For each hail event, find nearest station
def annotate_hail_with_station(hail_df, stations_df):
    results = []
    for _, row in hail_df.iterrows():
        closest = find_closest_station(row["Lat"], row["Lon"], stations_df)
        results.append({
            **row.to_dict(),
            "Closest Station": closest["station"],
            "Station Name": closest["name"],
            "Distance (km)": round(closest["distance_km"], 2)
        })
    return pd.DataFrame(results)
```

```python
import pdfplumber
import pandas as pd

pdf_path = "station_list/wsr88d-radar-list.pdf"  # replace with your file
output_csv = "station_list/nexrad_sites.csv"

rows = []

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        table = page.extract_table()
        if table:
            for row in table:
                rows.append(row)

# Convert to DataFrame
df = pd.DataFrame(rows)

# Optional: clean header if repeated or misaligned
df.columns = df.iloc[0]  # first row as header
df = df[1:]  # drop header row from data
```

```python
def find_closest_nexrad_id(hail_lat, hail_lon, site_csv_path="station_list/nexrad_sites.csv"):
    # Load NEXRAD site data
    sites = pd.read_csv(site_csv_path)

    # Clean column names
    sites.columns = [col.strip() for col in sites.columns]

    # Drop rows with missing coordinates
    sites = sites.dropna(subset=["LATITUDE_N", "LONGITUDE_W"])

    # Compute distance from hail location to each radar site
    sites["distance_km"] = sites.apply(
        lambda row: geodesic(
            (hail_lat, hail_lon), (row["LATITUDE_N"], -abs(row["LONGITUDE_W"]))
        ).kilometers,
        axis=1
    )

    # Return the ID of the closest site
    return sites.sort_values("distance_km").iloc[0]["ID"]
```

```python
import boto3
from botocore import UNSIGNED
from botocore.config import Config
import os
from datetime import datetime, timedelta

# Use anonymous access for public NEXRAD data
s3 = boto3.client("s3", config=Config(signature_version=UNSIGNED))
BUCKET_NAME = "noaa-nexrad-level2"


def download_radar_files(site_id, event_time_str, output_folder="radar_data"):
    os.makedirs(output_folder, exist_ok=True)

    if isinstance(event_time_str, str):
        event_time = datetime.strptime(event_time_str, "%Y-%m-%d %H:%M:%S")
    else:
        event_time = event_time_str
    start_time = event_time - timedelta(minutes=15)
    end_time = event_time + timedelta(minutes=15)

    # Format S3 prefix: YYYY/MM/DD/SITE_ID/
    prefix = f"{event_time.year:04}/{event_time.month:02}/{event_time.day:02}/{site_id}/"
    print(f"Looking in bucket: {BUCKET_NAME}/{prefix}")

    try:
        response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix=prefix)
        if "Contents" not in response:
            print("❌ No radar files found.")
            return

        count = 0
        for obj in response["Contents"]:
            key = obj["Key"]
            filename = key.split("/")[-1]
            try:
                file_time = datetime.strptime(filename[4:19], "%Y%m%d_%H%M%S").replace(tzinfo=pytz.UTC)
                if start_time <= file_time <= end_time:
                    out_path = os.path.join(output_folder, filename)
                    if not os.path.exists(out_path):
                        print(f"⬇️ Downloading: {filename}")
                        s3.download_file(BUCKET_NAME, key, out_path)
                        count += 1
                    else:
                        print(f"✔️ Already exists: {filename}")
            except Exception as parse_err:
                print(f"Skipping {filename}: {parse_err}")

        if count == 0:
            print("⚠️ No matching files in time range.")
    except Exception as e:
        print(f"❌ Error accessing S3: {e}")
```

```python
# UTC datetime of a hail event
event_time = datetime(2025, 7, 12, 22, 50, 0, tzinfo=pytz.UTC)

# Closest radar ID
radar_id = "KDVN"

# Download files
download_radar_files(radar_id, event_time)
```
